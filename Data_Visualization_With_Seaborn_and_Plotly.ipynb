```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
student = pd.read_csv('sns_data (1).csv')
student.head()
```

## Lineplot using Seaborn


```python
sns.lineplot(data= student, x = 'week', y = 'attendance_rate', errorbar = None, hue = 'gender')
```

```python
sns.relplot(kind ='line',data= student, x = 'week', y = 'attendance_rate', errorbar = None, hue = 'gender')
```

```python
sns.relplot(kind ='line',data= student, x = 'week', y = 'attendance_rate', errorbar = None, hue = 'class_level')
```

## Scatterplot using Seaborn

```python
sns.scatterplot(data = student, x = 'study_hours', y = 'test_score', hue = 'gender', style = 'class_level')
```

```python
sns.relplot(kind = 'scatter', data = student, x = 'study_hours', y = 'test_score', hue = 'gender', style = 'class_level')
```

```python
tips_data = sns.load_dataset("tips")
tips_data.head()
```

```python
sns.relplot(kind = 'scatter', data = tips_data, x = 'total_bill', y = 'tip', hue = 'sex', style = 'time', size = 'size')
```

## Facet Plot Using Seaborn

```python
student = pd.read_csv('student_dataset_complete.csv')
student
```

```python
sns.scatterplot(data= student, x = 'study_hours', y = 'test_score',hue = 'gender')
```

sns.relplot(kind = 'scatter', data= student, x = 'study_hours', y = 'test_score',col = 'gender')

## Histogram using Seaborn

```python
# axis lavel
sns.histplot(data = student, x = 'attendance_rate', hue = 'gender', bins = 15)
```

```python
# figure level
sns.displot(kind = 'hist', data = student, x = 'study_hours', col = 'gender')
```

```python
tips_data = sns.load_dataset("tips")
tips_data.head()
```

```python
sns.displot(kind = 'hist', data = tips_data, x = 'total_bill', bins = 20)
```

```python
sns.displot(kind = 'hist', data = tips_data, x = 'tip', bins = 20)
```

## KDE plot using Seaborn

```python
sns.kdeplot(data = student, x = 'attendance_rate')
```

```python
sns.kdeplot(data = tips_data, x = 'tip', hue = 'sex', fill = True)
```

##  Count Plot and Bar Plot using Seaborn

```python
sns.countplot(data = student, x = 'gender')
```

```python
sns.countplot(data = student, x = 'subject')
```

## Bar plot

```python
sns.barplot(data = student, x = 'gender', y = 'Marks', errorbar = None)
```

```python
sns.barplot(data = student, x = 'gender', y = 'Marks', errorbar = None, estimator = np.median)
```

## Regplot using seaborn

```python
student
```

```python
sns.regplot(data = student, x = 'study_hours', y = 'test_score')
```

```python
sns.lmplot(data = student, x = 'study_hours', y = 'test_score', hue = 'gender')
```

## Pairplot and Jointplot using Seaborn

```python
student_marks = student[['Marks', 'study_hours', 'attendance_rate', 'gender']]
student_marks
```

```python
sns.pairplot(data = student_marks)
```

```python
sns.pairplot(data = student_marks, kind = 'hist', hue = 'gender')
```

```python
sns.jointplot(data = student_marks, x = 'study_hours', y = 'Marks')
```

```python
sns.jointplot(data = student_marks, x = 'study_hours', y = 'Marks', kind = 'kde')
```

```python
sns.jointplot(data = student_marks, x = 'study_hours', y = 'Marks', hue = 'gender')
```
